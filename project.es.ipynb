#### Ejercicio 01. Análisis de datos básico con estructuras nativas de python.
Crea una lista con las calificaciones de 5 estudiantes: [8.5, 9.2, 7.8, 8.9, 10].

- Calcula el promedio de las calificaciones.

```python
# Código
grades = [8.5, 9.2, 7.8, 8.9, 10]
all_grades = 0

for notes in grades:
    all_grades += notes

all_grades /= len(grades)

print(f'El promedio de las calificaciones es: {all_grades:.2f}')
```

- Encuentra la calificación más alta y la más baja.

```python
# Código
max_grade = max(grades)
min_grade = min(grades)

print(f'La calificación más alta es: {max_grade:.2f}')
print(f'La calificación más baja es: {min_grade:.2f}')
```

#### Ejercicio 01. ¿Cuál es la casa más cara en todo el dataset? (★☆☆)

Imprime la dirección y el precio de la casa seleccionada. Por ejemplo:

`La casa con dirección en Calle del Prado, Nº20 es la más cara y su precio es de 5000000 USD`

```python
# TODO
#ordeno en un nuevo dataset por precio en orden descendente 
top_price_houses = ds.sort_values(by='price', ascending=False)
max_price_house = top_price_houses.iloc[0]
#print(max_price_house)
address = max_price_house['address']
price = max_price_house['price']
print(f'La casa con dirección en {address}, es la más cara y su precio es de {price:,} USD')
```

#### Ejercicio 02. ¿Cuál es la casa más barata del dataset? (★☆☆)

Este ejercicio es similar al ejercicio anterior, solo que ahora buscamos la casa con el precio más bajo. Recuerda imprimir la dirección y el precio de la casa seleccionada. Por ejemplo:

`La casa con dirección en Calle Alcalá, Nº58 es la más barata y su precio es de 12000 USD`

```python
# TODO
#porque tenemos casas con valor 0 y en alquier por menos de 1000 usd
price_house_clean = ds[ds['price'] > 1000]
#print(price_house_clean['price'].head())
bottom_price_houses = price_house_clean.sort_values(by='price')
min_price_house = bottom_price_houses.iloc[0]
#print(bottom_price_houses.head())
address = min_price_house['address']
price = min_price_house['price']
print(f'La casa con dirección en {address}, es la más barata y su precio es de {price:,} USD')
```

#### Ejercicio 07. ¿Cuál es la media de precios en la población de "Arroyomolinos (Madrid)"? (★★☆)

Imprima el valor obtenido de la columna level5.

```python
# TODO
filter_arroyomolinos = ds[ds['level5'] == 'Arroyomolinos (Madrid)']
mean_arroyomolinos = filter_arroyomolinos['price'].mean()
print(f'La media de precios en Arroyomolinos (Madrid) es: {mean_arroyomolinos:,.2f} USD')
```

#### Ejercicio 08. Trazar el histograma de los precios para la población de "Arroyomolinos (Madrid)" y explica qué observas (★★☆)

Imprime el histograma de los precios y escribe en la celda del Markdown un breve análisis del trazado.


```python
import matplotlib.pyplot as plt

# Trazar el histograma de precios
prices_arroyo = filter_arroyomolinos['price'] 

plt.figure(figsize=(10,5))
plt.hist(prices_arroyo, bins=20, color='mediumseagreen', edgecolor='black')

plt.title('Distribución de Precios en Arroyomolinos (Madrid)')
plt.xlabel('Precio (USD)')
plt.ylabel('Frecuencia (Número de casas)')
plt.show()
```
